```python
import pandas as pd
df = pd.read_csv('psws_design_charette_issues.csv', encoding='cp1252' )
df
```

```python
df.columns
```

```python
for index, row in df.iterrows():
    print(row['Name'], row['Suggested Feature Name'])
    with open("Issues/" + row['Suggested Feature Name'] + ".md", "w") as file_object:
        # Use the 'file' argument in print() to direct output to the file
        # print("This line will be written to the file.", file=file_object)
        # print("Another line for the file.", file=file_object)
        author = row['Name']
        email = row['Email']
        title = row['Suggested Feature Name']
        callsign = row['Callsign']
        description = row['Feature Description']
        comments = row['General Comments (optional)']
        print (f"""---
                name: Charette
                about: A charette suggestion
                title: "CHR: {title}"
                labels: charette
                ---
                
                # Description
                {description}
                
                ## Suggested by:
                {author}, {callsign}, {email}

                ## Additional comments:
                {comments}
                 """, file=file_object)
    
        # You can also print variables

        # print(f"Name: {name}, Age: {age}", file=file_object)

    print("Output has been written to Markdown file.")
    print(row['Name'], row['Suggested Feature Name'])
```
